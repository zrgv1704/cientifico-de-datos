```python
#cambiar fomrnato de fecha del dataframe fact_sales (ventas)
from datetime import datetime, timedelta

# Diccionario para traducir meses en español a inglés
meses_esp_en = {
    "ene": "Jan", "feb": "Feb", "mar": "Mar", "abr": "Apr",
    "may": "May", "jun": "Jun", "jul": "Jul", "ago": "Aug",
    "sep": "Sep", "oct": "Oct", "nov": "Nov", "dic": "Dec"
}

def convertir_fecha(fecha):
    partes = fecha.split('-')
    if len(partes[0]) <= 2:  # Formato semana-año
        semana = int(partes[0])
        año = int('20' + partes[1])
        fecha_inicio = datetime.strptime(f'{año} {semana} 1', '%Y %W %w')
        return fecha_inicio.date()
    else:  # Formato mes-año
        mes_esp = partes[0].lower()[:3]
        año = partes[1]
        mes_en = meses_esp_en.get(mes_esp)
        if mes_en:
            return datetime.strptime(f'{mes_en}-{año}', '%b-%y').date()
        else:
            return None


ventas['week_estandarizada'] = ventas['WEEK'].apply(convertir_fecha)
        
```

```python
# Unir dataframe productos y ventas
producto_venta = pd.concat([producto_1,ventas], axis=1)
```

```python
#Unir categorias y segmentos
categ_seg = pd.concat([categoria, segmento_1], axis=1)
categ_seg.head()
```

```python
#Unir ventas y categorias
df_consolidado = pd.concat([producto_venta,categ_seg], axis=1)
df_consolidado.head()
```

```python
#Agrupar por BRAND, segment, total_unit_sales
df_ventasf = df_consolidado.groupby(['BRAND','SEGMENT','TOTAL_UNIT_SALES'])
df_ventasf.head()
```

```python
#Calcular promedio de TOTAL_UNIT_SALES
promedio_ventas = df_ventasf['TOTAL_UNIT_SALES'].mean()
print(promedio_ventas)
```

```python
df_final = promedio_ventas.to_frame()
print(df_final)
```

```python
#Exportar a excel 
df_final.to_excel('datosConsolidados.xlsx', index=False)
```
